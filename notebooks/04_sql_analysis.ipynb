## Sales Analysis — Monthly Revenue Trend

```python
query = """
SELECT
    DATE_FORMAT(o.order_purchase_timestamp, '%Y-%m') AS month,
    ROUND(SUM(oi.price), 2) AS monthly_revenue
FROM orders o
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY DATE_FORMAT(o.order_purchase_timestamp, '%Y-%m')
ORDER BY month
"""

monthly_revenue = pd.read_sql(query, conn)
(print(monthly_revenue.head(5)))
```

## Revenue by Product Category

```python
query = """
SELECT
    COALESCE(p.product_category_name, 'Unknown') AS category,
    ROUND(SUM(oi.price), 2) AS revenue
FROM order_items oi
JOIN products p
    ON oi.product_id = p.product_id
GROUP BY p.product_category_name
ORDER BY revenue DESC
"""

category_revenue = pd.read_sql(query, conn)

print(category_revenue)
```

## Top-Selling Products

```python
query = """
SELECT
    oi.product_id,
    COUNT(*) AS units_sold,
    ROUND(SUM(oi.price), 2) AS revenue
FROM order_items oi
GROUP BY oi.product_id
ORDER BY units_sold DESC
LIMIT 10
"""

top_products = pd.read_sql(query, conn)

print(top_products)
```

## 3. Customer Analysis
Customer distribution
Customer spending
Repeat vs new customers
Top customers


```python
# 1. Customer Distribution
query_distribution = """
SELECT
    customer_state,
    COUNT(DISTINCT customer_unique_id) AS customers
FROM customers
GROUP BY customer_state
ORDER BY customers DESC
"""

customer_distribution = pd.read_sql(query_distribution, conn)

print("=== CUSTOMER DISTRIBUTION ===")
print(customer_distribution.head(5))


# 2. Customer Spending
query_spending = """
SELECT
    c.customer_unique_id,
    ROUND(SUM(oi.price), 2) AS total_spending
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY c.customer_unique_id
ORDER BY total_spending DESC
"""

customer_spending = pd.read_sql(query_spending, conn)

print("\n=== CUSTOMER SPENDING ===")
print(customer_spending.head(5))


# 3. Repeat vs New Customers
query_repeat = """
SELECT
    CASE
        WHEN order_count > 1 THEN 'Repeat Customer'
        ELSE 'New Customer'
    END AS customer_type,
    COUNT(*) AS customers
FROM (
    SELECT
        c.customer_unique_id,
        COUNT(DISTINCT o.order_id) AS order_count
    FROM customers c
    JOIN orders o
        ON c.customer_id = o.customer_id
    GROUP BY c.customer_unique_id
) AS customer_orders
GROUP BY customer_type
ORDER BY customers DESC
"""

repeat_customers = pd.read_sql(query_repeat, conn)

print("\n=== REPEAT VS NEW CUSTOMERS ===")
print(repeat_customers.head(5))


# 4. Top Customers
query_top = """
SELECT
    c.customer_unique_id,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(SUM(oi.price), 2) AS total_spending
FROM customers c
JOIN orders o
    ON c.customer_id = o.customer_id
JOIN order_items oi
    ON o.order_id = oi.order_id
GROUP BY c.customer_unique_id
ORDER BY total_spending DESC
LIMIT 10
"""

top_customers = pd.read_sql(query_top, conn)

print("\n=== TOP 5 CUSTOMERS ===")
print(top_customers.head(5))
```

## 4. Seller & Product Analysis
Top sellers
Seller revenue
Product/category performance
Seller ratings

```python
# 1. Top Sellers
query_top_sellers = """
SELECT
    seller_id,
    COUNT(*) AS units_sold
FROM order_items
GROUP BY seller_id
ORDER BY units_sold DESC
LIMIT 5
"""

top_sellers = pd.read_sql(query_top_sellers, conn)

print("=== TOP SELLERS ===")
print(top_sellers.head(5))


# 2. Seller Revenue
query_seller_revenue = """
SELECT
    seller_id,
    ROUND(SUM(price), 2) AS seller_revenue
FROM order_items
GROUP BY seller_id
ORDER BY seller_revenue DESC
LIMIT 5
"""

seller_revenue = pd.read_sql(query_seller_revenue, conn)

print("\n=== SELLER REVENUE ===")
print(seller_revenue.head(5))


# 3. Product / Category Performance
query_category_performance = """
SELECT
    p.product_category_name AS category,
    COUNT(oi.product_id) AS units_sold,
    ROUND(SUM(oi.price), 2) AS revenue
FROM order_items oi
JOIN products p
    ON oi.product_id = p.product_id
GROUP BY p.product_category_name
ORDER BY revenue DESC
LIMIT 5
"""

category_performance = pd.read_sql(query_category_performance, conn)

print("\n=== PRODUCT / CATEGORY PERFORMANCE ===")
print(category_performance.head(5))


# 4. Seller Ratings
query_seller_ratings = """
SELECT
    oi.seller_id,
    ROUND(AVG(r.review_score), 2) AS average_rating
FROM order_items oi
JOIN order_reviews r
    ON oi.order_id = r.order_id
GROUP BY oi.seller_id
HAVING COUNT(r.review_score) >= 5
ORDER BY average_rating DESC
LIMIT 5
"""

seller_ratings = pd.read_sql(query_seller_ratings, conn)

print("\n=== SELLER RATINGS ===")
print(seller_ratings.head(5))
```

## 5. Delivery Analysis
Average delivery time
On-time vs delayed orders
Delivery performance by location
Delivery delay vs review score

```python
# 1. Average Delivery Time
query_delivery_time = """
SELECT
    ROUND(AVG(
        TIMESTAMPDIFF(
            DAY,
            order_purchase_timestamp,
            order_delivered_customer_date
        )
    ), 2) AS average_delivery_days
FROM orders
WHERE order_delivered_customer_date IS NOT NULL
"""

delivery_time = pd.read_sql(query_delivery_time, conn)

print("=== AVERAGE DELIVERY TIME ===")
print(delivery_time.head(5))


# 2. On-Time vs Delayed Orders
query_delivery_status = """
SELECT
    CASE
        WHEN order_delivered_customer_date <= order_estimated_delivery_date
            THEN 'On Time'
        ELSE 'Delayed'
    END AS delivery_status,
    COUNT(*) AS total_orders
FROM orders
WHERE order_delivered_customer_date IS NOT NULL
  AND order_estimated_delivery_date IS NOT NULL
GROUP BY delivery_status
ORDER BY total_orders DESC
"""

delivery_status = pd.read_sql(query_delivery_status, conn)

print("\n=== ON-TIME VS DELAYED ORDERS ===")
print(delivery_status.head(5))


# 3. Delivery Performance by Location
query_location_delivery = """
SELECT
    c.customer_state,
    COUNT(DISTINCT o.order_id) AS total_orders,
    ROUND(AVG(
        TIMESTAMPDIFF(
            DAY,
            o.order_purchase_timestamp,
            o.order_delivered_customer_date
        )
    ), 2) AS average_delivery_days
FROM orders o
JOIN customers c
    ON o.customer_id = c.customer_id
WHERE o.order_delivered_customer_date IS NOT NULL
GROUP BY c.customer_state
ORDER BY average_delivery_days
"""

location_delivery = pd.read_sql(query_location_delivery, conn)

print("\n=== DELIVERY PERFORMANCE BY LOCATION ===")
print(location_delivery.head(5))


# 4. Delivery Delay vs Review Score
query_delay_review = """
SELECT
    r.review_score,
    ROUND(AVG(
        DATEDIFF(
            o.order_delivered_customer_date,
            o.order_estimated_delivery_date
        )
    ), 2) AS average_delivery_delay
FROM orders o
JOIN order_reviews r
    ON o.order_id = r.order_id
WHERE o.order_delivered_customer_date IS NOT NULL
  AND o.order_estimated_delivery_date IS NOT NULL
GROUP BY r.review_score
ORDER BY r.review_score
"""

delay_review = pd.read_sql(query_delay_review, conn)

print("\n=== DELIVERY DELAY VS REVIEW SCORE ===")
print(delay_review.head(5))
```

## 6. Customer Experience
Review score distribution
Reviews by category
Rating vs delivery performance


```python
# 1. Review Score Distribution
query_review_distribution = """
SELECT
    review_score,
    COUNT(*) AS total_reviews
FROM order_reviews
GROUP BY review_score
ORDER BY review_score
"""

review_distribution = pd.read_sql(query_review_distribution, conn)

print("=== REVIEW SCORE DISTRIBUTION ===")
print(review_distribution.head(5))


# 2. Reviews by Category
query_reviews_category = """
SELECT
    p.product_category_name AS category,
    COUNT(r.review_id) AS total_reviews,
    ROUND(AVG(r.review_score), 2) AS average_rating
FROM order_reviews r
JOIN order_items oi
    ON r.order_id = oi.order_id
JOIN products p
    ON oi.product_id = p.product_id
WHERE p.product_category_name IS NOT NULL
GROUP BY p.product_category_name
ORDER BY total_reviews DESC
"""

reviews_category = pd.read_sql(query_reviews_category, conn)

print("\n=== REVIEWS BY CATEGORY ===")
print(reviews_category.head(5))


# 3. Rating vs Delivery Performance
query_rating_delivery = """
SELECT
    r.review_score,
    COUNT(*) AS total_reviews,
    ROUND(AVG(
        DATEDIFF(
            o.order_delivered_customer_date,
            o.order_estimated_delivery_date
        )
    ), 2) AS average_delivery_delay
FROM order_reviews r
JOIN orders o
    ON r.order_id = o.order_id
WHERE o.order_delivered_customer_date IS NOT NULL
  AND o.order_estimated_delivery_date IS NOT NULL
GROUP BY r.review_score
ORDER BY r.review_score
"""

rating_delivery = pd.read_sql(query_rating_delivery, conn)

print("\n=== RATING VS DELIVERY PERFORMANCE ===")
print(rating_delivery.head(5))
```
